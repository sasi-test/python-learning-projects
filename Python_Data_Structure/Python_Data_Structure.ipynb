```python
my_test_list = []
print("Let's add 3 items to your list!")
for i in range(3):
    new_item = input(f"Enter item{i+1}:")
    my_test_list.append(new_item)
print("\n-- Here is your final list from runtime---")
print(my_test_list)
```

```python
# Create a blank list to store our messages/items
my_test_list = []

print("=== Chat-Style List Builder ===")
print("Type your messages one by one.")
print("Type 'done' when you want to finish and see your list.\n")

# Use a while loop to keep taking input until the user says 'done'
while True:
    new_item = input("Enter a message: ")
    
    # Check if the user wants to stop
    if new_item.lower() == 'done':
        break  # Exits the loop immediately
        
    # Append the typed message into our list
    my_test_list.append(new_item)

# Print the final collected list
print("\n--- Here is your final collected list ---")
print(my_test_list)
```

## Learning abt set, set is used to remove duplicated 
* wont allow duplicate value
* hetrogenius
* 1 dimension array
* indexing and slicing is not possible 

```python
# 1. LIST: Collecting incoming items dynamically (allows duplicates and updates)
print("--- Step 1: Collecting incoming orders (List) ---")
incoming_orders = ["Laptop", "Mouse", "Keyboard", "Mouse", "Monitor"]
incoming_orders[1] = "Mouse Pad"

print("Orders received:", incoming_orders)

# 2. TUPLE: Store fixed store info or configuration settings (cannot be accidentally changed)
store_info = ("TechHub Store", "New York", 2026)
print("\n--- Step 2: Store Details (Tuple) ---")
print(f"Store: {store_info[0]} | Location: {store_info[1]} | Established: {store_info[2]}")

# 3. SET: Cleaning the orders to see unique products sold today (removes duplicates)
print("\n--- Step 3: Unique Products Sold (Set) ---")
unique_products = set(incoming_orders)
print("Unique items sold today:", unique_products)
```

## Learning abt dictonary, this is a key value pair can do any operation mentioned above but can access ony using this key not by index.

```python
# 1. Create a blank dictionary
student_info = {}

# 2. Get key and value from the user at runtime
user_key = input("Enter the detail you want to store (e.g., name, age, city): ")
user_value = input(f"Enter the value for {user_key}: ")

# Store it in the dictionary using the key
student_info[user_key] = user_value

# 3. Retrieve the value back out using that key at runtime
print("\n--- Retrieving Data ---")
print(f"You searched for '{user_key}', and the value is:", student_info[user_key])
```

```python
student_marks = {
    "Name": ["sasi", "Vimala", "Manjula","Akash"], 
    "Math": [90, 70, 100, 50],
    "Science": [85, 80, 95, 80]
}

# zip() can take 3 (or even more) lists at once!
for name, math_score, sci_score in zip(student_marks["Name"], student_marks["Math"], student_marks["Science"]):
    print(f"Student: {name} | Math: {math_score} | Science: {sci_score}")
```

### About indexing: index stars from 0. Indexing from last starts from -1. 

```python
# A list of daily temperature readings over a week
weekly_temperatures = [32, 34, 31, 35, 33, 36, 38]

# Using positive indexing to get the first day
print("First day's temperature (Index 0):", weekly_temperatures[0])

# Using negative indexing to get the absolute latest (most recent) day
print("Latest temperature today (Index -1):", weekly_temperatures[-1])

# Using negative indexing to get yesterday's temperature
print("Yesterday's temperature (Index -2):", weekly_temperatures[-2])
```
