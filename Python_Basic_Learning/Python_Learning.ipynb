## Checking if the user is entering string instead of no

```python
import random

Secret_no = random.randint(1, 100)
print("Secret_no:", Secret_no)

print("Guess My Number")
print("Enter the no between 1 to 100:")

while True:
    # 1. Get input as a string first
    user_input = input("Enter the no: ")
    
    # 2. Check if it's NOT a digit
    if not user_input.isdigit():
        print("Invalid input! Please enter a number between 1 to 100:")
        continue  # Skips the rest and asks again
        
    # 3. Convert to integer now that it's safe
    user_guessed_no = int(user_input)
    
    # 4. Game conditions (aligned with proper 4-space indentation)
    if user_guessed_no < 1 or user_guessed_no > 100:
        print("Invalid no")
    elif user_guessed_no > Secret_no:
        print("The no you have entered is greater than the secret no.")
    elif user_guessed_no < Secret_no:
        print("The no you have entered is less than the secret no.")
    else:
        print("You got it right!")
        break
```

## Testing the same function with While loop with condition
