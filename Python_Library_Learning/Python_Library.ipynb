```python
import pandas as pd
import numpy as np

# 1. Read your file
Testing_File = pd.read_csv("sample_test_data.csv")

# 2. FILTER & CLEAN: Strip the "s" and convert to numbers (float)
raw_times = Testing_File["Execution_Time"].str.replace("s", "").astype(float)
#print(np.rep)
# 3. Send the clean numbers into NumPy
np_times = np.array(raw_times)

print("Cleaned NumPy Array:", np_times)
print("Average Execution Time:", np.mean(np_times))
print("Minimum Executtion Time:", np. min(np_times))
```

## Learning abt matplotlib 

```python
import matplotlib.pyplot as plt

# 1. Grab your test names from your DataFrame for the X-axis
test_names = Testing_File["Test_Name"]

# 2. Use the clean NumPy array of execution times we made earlier for the Y-axis
execution_times = np_times

# 3. Create a bar chart
#plt.bar(test_names, execution_times, color="skyblue")
# Line chat
#plt.plot(test_names, execution_times, marker='o', color='green', linewidth=2)
#Bar chat
#plt.barh(test_names, execution_times, color="orange")
#Scatter chat
plt.scatter(test_names, execution_times, color="purple", s=100)

# 4. Add labels and a title to make it a professional report chart
plt.xlabel("Test Cases")
plt.ylabel("Execution Time (seconds)")
plt.title("Test Execution Time Report")

# 5. Display the chart
plt.show()
```
